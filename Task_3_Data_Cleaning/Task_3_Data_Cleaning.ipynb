# TASK 3 - DATA CLEANING

## 1. Loading the Dataset

In this project, we are using the Titanic dataset.
The objective is to identify and fix data quality issues
and prepare a clean dataset for analysis.

```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("train.csv")

print("Dataset Loaded Successfully!")
```

```python
display(df.head(10))
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
print("Missing Values in Each Column:")

print(df.isnull().sum())
```

```python
print("Total Duplicate Rows:", df.duplicated().sum())
```

## 2. Data Quality Report

This section identifies missing values, duplicate records,
data types and unusual numerical values in the dataset.

```python
quality_report = pd.DataFrame({
    'Column Name': df.columns,
    'Missing Values': df.isnull().sum().values,
    'Data Type': df.dtypes.values,
    'Unique Values': df.nunique().values
})

display(quality_report)
```

## 3. Missing Value Handling

Missing values can affect data analysis.
In this section, we will identify missing values
and handle them using appropriate techniques.

```python
print("Missing Values Before Cleaning:")
print(df.isnull().sum())
```

```python
# Fill missing Age using median
df['Age'] = df['Age'].fillna(df['Age'].median())

# Fill missing Embarked using mode
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Drop Cabin column
df.drop(columns=['Cabin'], inplace=True)

print("Missing Values Handled Successfully!")
```

```python
print("Missing Values After Cleaning:")
print(df.isnull().sum())
```

## 4. Duplicate Removal

Duplicate records can affect the accuracy of data analysis.
We will identify and remove duplicate rows.

```python
# Count duplicates before removal
duplicates_before = df.duplicated().sum()

print("Duplicates Before Cleaning:", duplicates_before)

# Remove duplicates
df.drop_duplicates(inplace=True)

# Count duplicates after removal
duplicates_after = df.duplicated().sum()

print("Duplicates After Cleaning:", duplicates_after)

print("Total Duplicates Removed:", duplicates_before - duplicates_after)
```

## 5. Data Standardisation

Data standardisation ensures consistent formatting
across categorical columns.

```python
# Standardise gender values
df['Sex'] = df['Sex'].str.strip().str.lower()

df['Sex'] = df['Sex'].replace({
    'male': 'Male',
    'female': 'Female'
})

# Remove extra spaces from Embarked
df['Embarked'] = df['Embarked'].str.strip().str.upper()

print("Data Standardisation Completed!")

print(df['Sex'].value_counts())
print(df['Embarked'].value_counts())
```

```python
print("Final Missing Values:")
print(df.isnull().sum())

print("\nFinal Duplicate Count:")
print(df.duplicated().sum())
```

## 6. Outlier Detection

In this section, we use the IQR (Interquartile Range)
method to identify unusual values in numerical columns.

```python
# Select numerical columns
numeric_columns = df.select_dtypes(include=np.number).columns

# Detect outliers using IQR
for col in numeric_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    print(f"{col}: {len(outliers)} outliers")
```

## 7. Data Type Correction

In this section, we correct the data types of columns
to ensure the dataset is suitable for further analysis.

```python
print("Before Data Type Correction:")
print(df.dtypes)
```

```python
# Convert PassengerId into string
df['PassengerId'] = df['PassengerId'].astype(str)

# Convert Age into float
df['Age'] = df['Age'].astype(float)

# Convert Fare into float
df['Fare'] = df['Fare'].astype(float)

# Convert Survived into integer
df['Survived'] = df['Survived'].astype(int)

print("Data Types Corrected Successfully!")
```

```python
print("After Data Type Correction:")
print(df.dtypes)
```

## 8. Before vs After Cleaning Summary

In this section, we compare the original dataset
with the cleaned dataset to evaluate improvements
in data quality.

```python
# Load original dataset for comparison
original_df = pd.read_csv("train.csv")

print("Original Dataset Loaded Successfully!")
```

```python
# Expected data types
expected_dtypes = {
    'PassengerId': 'object',
    'Survived': 'int64',
    'Pclass': 'int64',
    'Name': 'object',
    'Sex': 'object',
    'Age': 'float64',
    'SibSp': 'int64',
    'Parch': 'int64',
    'Ticket': 'object',
    'Fare': 'float64',
    'Embarked': 'object'
}

# Calculate dtype accuracy
def dtype_accuracy(data):
    correct = sum(
        str(data[col].dtype) == dtype
        for col, dtype in expected_dtypes.items()
        if col in data.columns
    )
    return f"{correct}/{len(expected_dtypes)}"

# Create summary
summary = pd.DataFrame({
    'Metric': [
        'Total Rows',
        'Missing Values',
        'Duplicate Rows',
        'Total Columns',
        'Data Type Accuracy'
    ],

    'Before Cleaning': [
        original_df.shape[0],
        original_df.isnull().sum().sum(),
        original_df.duplicated().sum(),
        original_df.shape[1],
        dtype_accuracy(original_df)
    ],

    'After Cleaning': [
        df.shape[0],
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        df.shape[1],
        dtype_accuracy(df)
    ]
})

display(summary)
```
